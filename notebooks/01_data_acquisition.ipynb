# 01 — Data Acquisition

This notebook is responsible for acquiring the raw ADMET data used in the project.

The data source is the **Therapeutics Data Commons (TDC)**, using the official **ADMET Benchmark Group**. At this stage, the goal is only to download and load the original benchmark data without applying any cleaning, filtering, transformation, or feature engineering. 

The first dataset used in the project is **Solubility_AqSolDB**, which contains molecular structures and experimental solubility values.

## Objectives

- Initialize the TDC ADMET benchmark group.
- Download and load the `Solubility_AqSolDB` benchmark.
- Preserve the original train/validation and test partitions provided by TDC.
- Store the raw data locally for use in later stages of the project.

```python
from pathlib import Path

from tdc.benchmark_group import admet_group
```

```python
group = admet_group(path="../data/raw/")
benchmark = group.get("Solubility_AqSolDB")

# First dataset we will approach is the Solubility_AqSolDB benchmark, 
# which contains 9980 molecules with their solubility values in water. 
# The benchmark is split into a training + validation set and a test set.

print("Benchmark:", benchmark["name"])

benchmark["train_val"].head()
```
